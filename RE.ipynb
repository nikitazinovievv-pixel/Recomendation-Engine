```python
import pandas as pd
import numpy as np 
import pathlib as pl
import sklearn 
from sklearn import linear_model
from  sentence_transformers import SentenceTransformer
```

```python
N_path = pl.Path('/Users/nikitazinovev/Desktop/Study/Semester 3:4/Tools/Projects/netflix_titles.csv')
N_data = pd.read_csv(N_path)
N_data = N_data.dropna()

# Replace full country names to compact ones

N_data = N_data.replace('United States', 'USA', regex= True)
N_data = N_data.replace('United Kingdom', 'UK', regex= True)
N_data = N_data.replace('United Arab Emirates', 'UAE', regex= True)
N_data = N_data.replace('South Africa', 'SA', regex= True)
N_data = N_data.replace('South Korea', 'SK', regex= True)
N_data = N_data.replace('Hong Kong', 'HK', regex= True)
N_data = N_data.replace('New Zealand', 'NZ', regex= True)


display(N_data)
```

```python
model = SentenceTransformer('all-MiniLM-L6-v2')
description_embeddings = model.encode(N_data['country'].tolist(), show_progress_bar=True)

# Output shape will be: (number_of_titles, 384)
print(description_embeddings.shape)
```
